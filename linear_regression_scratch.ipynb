```python
import numpy as np
import matplotlib.pyplot as plt
x=np.array([21,24,30,40,50])
y=np.array([10,20,30,50,100])
x_mean=np.mean(x)
y_mean=np.mean(y)
nr=np.sum((x-x_mean)*(y-y_mean))
dr=np.sum((x-x_mean)**2)
m=nr/dr
c=y_mean-m*x_mean
print(f"Line equation y={m:.2f}x+{c:.2f}")
y_pred=m*x+c
plt.scatter(x,y,color='blue',label='actual')
plt.plot(x,y_pred,color='red',label='regression line')
plt.xlabel("diameter(inches)")
plt.ylabel("price($100s)")
plt.title("simple linear regression -from scratch")
plt.legend()
plt.grid(True)
plt.show()
new_area=150
predicted_price=m*new_area+c
print(f"predicted price for {new_area}inches :${predicted_price:.2f}k")





```
